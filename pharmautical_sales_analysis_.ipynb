```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
drug_columns = [
    'M01AB', 'M01AE', 'N02BA', 'N02BE',
    'N05B', 'N05C', 'R03', 'R06'
]
```

```python
total_sales = df[drug_columns].sum()

print(total_sales)
```

```python
total_sales = total_sales.sort_values(ascending=False)

print(total_sales)
```

```python
top_3_drugs = total_sales.head(3)
print(top_3_drugs)
```

```python
top_3_drugs.plot(kind='bar')

plt.title('Top 3 Drugs by Total Sales')
plt.xlabel('Drug Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
jan_2015 = df[(df['Year'] == 2015) & (df['Month'] == 1)]

jan_sales = jan_2015[drug_columns].sum().sort_values(ascending=False)

print("January 2015:")
print(jan_sales.head(3))
```

```python
july_2016 = df[(df['Year'] == 2016) & (df['Month'] == 7)]

july_sales = july_2016[drug_columns].sum().sort_values(ascending=False)

print("July 2016:")
print(july_sales.head(3))
```

```python
sep_2017 = df[(df['Year'] == 2017) & (df['Month'] == 9)]

sep_sales = sep_2017[drug_columns].sum().sort_values(ascending=False)

print("September 2017:")
print(sep_sales.head(3))
```

```python
sales_2017 = df[df['Year'] == 2017]
```

```python
sales_2017_total = sales_2017[drug_columns].sum().sort_values(ascending=False)

print(sales_2017_total)
```

```python
top_drug_2017 = sales_2017_total.head(1)

print("Drug with highest sales in 2017:")
print(top_drug_2017)
```

```python
average_sales = df[drug_columns].mean().sort_values(ascending=False)

print(average_sales)
```

```python
top_average_drug =average_sales.head(1)
print("Drug with highest average sales:")
print(top_average_drug)
```

```python
average_sales.plot(kind='bar')

plt.title('Average Daily Sales by Drug Category')
plt.xlabel('Drug Category')
plt.ylabel('Average Daily Sales')
plt.xticks(rotation=45)
plt.show()
```
